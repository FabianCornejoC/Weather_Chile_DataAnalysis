## Conclusiones

El análisis fue realizado utilizando datos públicos oficiales de Chile, proporcionados por el Instituto Nacional de Estadísticas (INE) e importados por la Dirección General de Aeronáutica Civil (DGAC). Los resultados permiten identificar tendencias climáticas regionales observadas en los últimos años.

A partir del análisis se concluye que:

- Los días con precipitación en la zona central del país presentan, en general, temperaturas máximas entre 10 °C y 20 °C, y temperaturas mínimas entre 5 °C y 14 °C.
- Se observa una tendencia negativa en la cantidad anual de días de lluvia en la zona central de Chile.
- Esta tendencia no se presenta de forma consistente en las demás zonas del país.
- El período cercano al año 2019 muestra el mayor déficit de precipitaciones, con valores que oscilan entre un 70 % y 90 % dependiendo de la región.

Estos resultados evidencian diferencias significativas en los patrones de precipitación entre las distintas zonas del país.

## Recomendaciones

- Ampliar el análisis incorporando variables adicionales, como factores geográficos o índices climáticos de largo plazo, que permitan profundizar en las posibles causas de la disminución de precipitaciones en la zona central.
- Extender el período de estudio incluyendo datos más recientes para evaluar si las tendencias observadas se mantienen en el tiempo.
- Realizar un análisis segmentado por región y estación del año para identificar patrones climáticos más específicos a nivel local.
